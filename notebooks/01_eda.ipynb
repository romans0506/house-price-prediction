```python
def compute_cost(X,y,w,b):
    m = len(y)
    

    prediction = np.dot(X, w) + b

    cost = (1 / (2 * m)) * np.sum((prediction - y)** 2)

    return cost
```

```python
def compute_gradient(X,y,w,b):


    m = len(X)

    dj_dw = 0
    dj_db = 0

    prediction = np.dot(X, w) + b

    error = prediction - y

    dj_dw = (1 / m) * np.dot(error, X)
    dj_db = (1 / m) * np.sum(error)

    return dj_dw, dj_db
```

```python
dj_dw, dj_db = compute_gradient(X_train_scaled, y_train_np,w, b)

print(dj_dw.shape)
print(dj_db)
```

```python
print(len(X_train_scaled))
```

## Doing gradient descent steps

```python
initial_cost = compute_cost(X_train_scaled,y_train_np,w,b)

print(initial_cost)
```

```python
def gradient_descent(X,y,w,b,alpha,num_iter):
    J_history = []

    for i in range(num_iter):

        dj_dw , dj_db = compute_gradient(X,y,w,b)

        w = w - alpha * dj_dw
        b = b - alpha * dj_db


        cost = compute_cost(X,y,w,b)
        J_history.append(cost)

        if i % 100 == 0:
            print(f"Iteration {i}: Cost {cost}")

    return w, b, J_history
```

```python
alpha = 0.001

w_init = np.zeros(X_train_scaled.shape[1])
b_init = 0.0
```

```python
w, b, J_history = gradient_descent(
    X_train_scaled,
    y_train_np,
    w_init,
    b_init,
    alpha,
    10000
)
```

```python
plt.plot(J_history)
plt.xlabel("Iteration")
plt.ylabel("Cost")
plt.title("Cost vs. Iteration")
plt.show()
```

## We used the Gradient Descent with the learning rate of 0.001 and it works pretty well
